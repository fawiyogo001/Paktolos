<h2>Pie Chart</h2>
<p>What are the top spending categories?</p>
<p> --> To display the Top 7 Categories of Expenses with the rest of the Categories grouped as "Others" to show priority areas</p>

```python
# Top 7 Expensive Categories
TOP_N = 7
CATEGORY_SUM = CASH_FLOW[CASH_FLOW['Income/Expenses'] == "Expenses"].groupby(["Category"])["Amount"].sum().reset_index().sort_values(by="Amount", ascending=False, ignore_index=True)
CATEGORY_SUM
```

```python
# Separate Top N Categories
TOP_CATEGORIES = CATEGORY_SUM.head(TOP_N)
OTHERS_TOTAL = CATEGORY_SUM.iloc[TOP_N:]["Amount"].sum()

OTHERS_ROW = pd.DataFrame({"Category": ["Others"], "Amount": [OTHERS_TOTAL]})

PIE_DATA = pd.concat([TOP_CATEGORIES, OTHERS_ROW], ignore_index=True)
PIE_DATA.columns = ["Category", "Amount"]
PIE_DATA["Allocation"] = round((PIE_DATA["Amount"] / PIE_DATA["Amount"].sum()) * 100, 2)
PIE_DATA
```

```python
# Pie Chart
fig = px.pie(
    PIE_DATA,
    names="Category",
    values="Amount",
    title=f"Spending by Category (Top {TOP_N} + Others)",
    template="plotly_dark",
    color_discrete_sequence=px.colors.qualitative.Pastel
)

fig.update_traces(textinfo="label+percent", textposition="outside")
fig.update_layout(legend_title_text="Category")

fig.show()
```

```python
# Top N Categories Allocation Data Summary
TOP_N_ALLOCATION = PIE_DATA[PIE_DATA["Category"] != "Others"]["Allocation"].sum()
TOP_N_HIGHEST_ALLOCATION = PIE_DATA[PIE_DATA["Category"] != "Others"].sort_values("Allocation", ascending=False).iloc[0]

TOP_N_HIGHEST_CATEGORY = TOP_N_HIGHEST_ALLOCATION["Category"]
TOP_N_HIGHEST_ALLOCATION = TOP_N_HIGHEST_ALLOCATION["Allocation"]

FIG_TITLE = fig.layout.title.text if fig.layout.title else "Untitled Plot"
PLOT_CAPTION = f"""
Top {TOP_N} categories account for {TOP_N_ALLOCATION:.2f}% of the total spending with the highest category spending being {TOP_N_HIGHEST_CATEGORY} at {TOP_N_HIGHEST_ALLOCATION:.2f}%.
"""

PLOT_CAPTION_WRAPPED = textwrap.fill(PLOT_CAPTION, width=130)
print(PLOT_CAPTION_WRAPPED)

print("\n--------------------------------\n")

SAVE_PLOT_AND_REGISTER(
    fig,
    FILENAME=f"PIE_CHART_TOP_{TOP_N}_SPENDING_CATEGORIES.png",
    TITLE=FIG_TITLE,
    CAPTION=PLOT_CAPTION_WRAPPED
)
```

```python
# Day with Highest Spending
DAY_WITH_HIGHEST_SPENDING = DAY_IN_A_WEEK_AGG.sort_values("AVERAGE_AMOUNT", ascending=False).iloc[0]
DAY_WITH_LOWEST_SPENDING = DAY_IN_A_WEEK_AGG.sort_values("AVERAGE_AMOUNT", ascending=True).iloc[0]
DAY_IN_A_WEEK_WITH_HIGHEST_SPENDING = DAY_WITH_HIGHEST_SPENDING["Day"]
DAY_IN_A_WEEK_WITH_LOWEST_SPENDING = DAY_WITH_LOWEST_SPENDING["Day"]

print(DAY_IN_A_WEEK_WITH_LOWEST_SPENDING, DAY_IN_A_WEEK_WITH_HIGHEST_SPENDING)
```

```python
# Average Gap between Day with Highest Spending and Rest of the Week
REST_OF_THE_WEEK = DAY_IN_A_WEEK_AGG[DAY_IN_A_WEEK_AGG["Day"] != DAY_IN_A_WEEK_WITH_HIGHEST_SPENDING]
AVERAGE_TOTAL_AMOUNT_OF_REST_OF_THE_WEEK = REST_OF_THE_WEEK["AVERAGE_AMOUNT"].mean()
AVERAGE_TOTAL_AMOUNT_OF_REST_OF_THE_WEEK

AVERAGE_GAP = DAY_WITH_HIGHEST_SPENDING["AVERAGE_AMOUNT"] - AVERAGE_TOTAL_AMOUNT_OF_REST_OF_THE_WEEK
AVERAGE_GAP_PERCENT = round((AVERAGE_GAP / AVERAGE_TOTAL_AMOUNT_OF_REST_OF_THE_WEEK) * 100, 1)
AVERAGE_GAP_PERCENT
```

```python
# Pivot Table for Surplus/Deficit Analysis 

MONTHLY_CASH_FLOW_PIVOT = MONTHLY_CASH_FLOW.pivot_table(
    index=["Year Month", "Year Month Label"],
    columns=["Income/Expenses", "Real Income"],
    values="Amount",
    aggfunc="sum"
).fillna(0).reset_index()
MONTHLY_CASH_FLOW_PIVOT

MONTHLY_CASH_FLOW_PIVOT.columns = ["Year Month", "Year Month Label", "Expenses", "Carry-over", "Real Income"]
MONTHLY_CASH_FLOW_PIVOT = MONTHLY_CASH_FLOW_PIVOT.rename(columns={
    ("Income", True): "Carry-over",
    ("Income", False): "Real Income",
    ("Expenses", "Expenses"): "Expenses"
})
MONTHLY_CASH_FLOW_PIVOT.head()
```

```python
# Cumulative Surplus
MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE = MONTHLY_CASH_FLOW_PIVOT.sort_values("Year Month").copy()
MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE["Carryover"] = 0.0
MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE["Surplus"] = 0.0
MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE["Cumulative Surplus"] = 0.0

for i in range(len(MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE)):
    CARRYOVER = MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.iloc[i - 1]["Surplus"] if i > 0 else 0.0
    MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.at[i, "Carryover"] = CARRYOVER
    PREV_CUMULATIVE_SURPLUS = MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.iloc[i - 1]["Cumulative Surplus"] if i > 0 else 0.0
    
    REAL_INCOME = MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.at[i, "Real Income"]
    EXPENSES_TOTAL_MONTHLY = MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.at[i, "Expenses"]
    
    # Ensure these are not NaN
    REAL_INCOME = 0.0 if pd.isna(REAL_INCOME) else REAL_INCOME
    EXPENSES_TOTAL_MONTHLY = 0.0 if pd.isna(EXPENSES_TOTAL_MONTHLY) else EXPENSES_TOTAL_MONTHLY
    
    SURPLUS = REAL_INCOME + CARRYOVER - EXPENSES_TOTAL_MONTHLY
    MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.at[i, "Surplus"] = SURPLUS
    MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.at[i, "Cumulative Surplus"] = SURPLUS + PREV_CUMULATIVE_SURPLUS

MONTHLY_CASH_FLOW_PIVOT_CUMULATIVE.head()
```

```python
# YoY Pivot Table
YOY_CASH_FLOW_PIVOT = YOY_CASH_FLOW.pivot_table(
    index="Month",
    columns="Year",
    values="Surplus",
    aggfunc="sum"
).sort_index()

YOY_CASH_FLOW_PIVOT = YOY_CASH_FLOW_PIVOT.reset_index()
YOY_CASH_FLOW_PIVOT["Month Label"] = YOY_CASH_FLOW_PIVOT["Month"].apply(lambda MONTH: calendar.month_name[MONTH])

YOY_CASH_FLOW_PIVOT
```

```python
MONTH_WITH_LOWEST_SURPLUS = YOY_CASH_FLOW.groupby(["Month"]).agg({
    "Surplus": "mean",
}).reset_index().sort_values("Surplus", ascending=True).head(3)
MONTH_WITH_LOWEST_SURPLUS_TOP_3 = MONTH_WITH_LOWEST_SURPLUS["Month"].tolist()
MONTH_WITH_LOWEST_SURPLUS_TOP_3 = [calendar.month_name[i] for i in sorted(MONTH_WITH_LOWEST_SURPLUS_TOP_3)]
MONTH_WITH_LOWEST_SURPLUS_TOP_3
```

```python
# SARIMA Model
SARIMA_MODEL = SARIMAX(
    MONTHLY_EXPENSES_TIME_SERIES,
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

SARIMA_MODEL = SARIMA_MODEL.fit(disp=False)
SARIMA_RESIDUALS = SARIMA_MODEL.resid
SARIMA_RESIDUALS = SARIMA_RESIDUALS.dropna()
```

```python
# SARIMA Forecasting for Early Warning (24 Months)
FORECAST_HORIZON = 24
SARIMA_CI_ALPHA = 0.05
SARIMA_FORECAST = SARIMA_MODEL.get_forecast(steps=FORECAST_HORIZON)

SARIMA_FORECAST_MEAN = SARIMA_FORECAST.predicted_mean
SARIMA_FORECAST_CI = SARIMA_FORECAST.conf_int(alpha=SARIMA_CI_ALPHA) 

SARIMA_FORECAST_LOWER_CI = SARIMA_FORECAST_CI.iloc[:, 0]
SARIMA_FORECAST_UPPER_CI = SARIMA_FORECAST_CI.iloc[:, 1]
```

```python
# SARIMA Forecast Visualisation
fig_layout = go.Layout(
    autosize=True,
    width=1000,
    height=700
)

fig = go.Figure(layout=fig_layout)

# Historical
fig.add_trace(go.Scatter(
    x=MONTHLY_EXPENSES_TIME_SERIES.index,
    y=MONTHLY_EXPENSES_TIME_SERIES,
    mode="lines",
    name="Historical Expenses",
    line=dict(color="#5DADE2")
))

# Forecast mean
fig.add_trace(go.Scatter(
    x=SARIMA_FORECAST_MEAN.index,
    y=SARIMA_FORECAST_MEAN,
    mode="lines",
    name="Forecast",
    line=dict(dash="dash", color="#F39C12")
))

# Confidence interval
fig.add_trace(go.Scatter(
    x=SARIMA_FORECAST_CI.index.tolist() + SARIMA_FORECAST_CI.index[::-1].tolist(),
    y=SARIMA_FORECAST_CI.iloc[:, 0].tolist() + SARIMA_FORECAST_CI.iloc[:, 1][::-1].tolist(),
    fill="toself",
    fillcolor="rgba(255,255,255,0.15)",
    line=dict(color="rgba(255,255,255,0)"),
    name="Confidence Interval"
))

fig.update_layout(
    title="SARIMA Forecast of Monthly Expenses",
    xaxis_title="Date",
    yaxis_title="Expenses",
    template="plotly_dark",
    height=600
)

fig.show()

FIG_TITLE = fig.layout.title.text if fig.layout.title else "Untitled Plot"
PLOT_CAPTION = f"""
Given historical data and model assumptions, there is a {(1 - SARIMA_CI_ALPHA) * 100}% probability that actual expenses will fall within this range. If an actual monthly expense exceeds the upper bound, it signals potential overspending requiring attention, while staying below the lower bound indicates underspending. The widening confidence interval reflects increasing uncertainty over longer forecast horizons, underscoring the need for regular forecast updates as new data becomes available.
"""

PLOT_CAPTION_WRAPPED = textwrap.fill(PLOT_CAPTION, width=130)
print(PLOT_CAPTION_WRAPPED)

print("\n--------------------------------\n")

SAVE_PLOT_AND_REGISTER(
    fig,
    FILENAME="SARIMA_FORECAST_FOUNDATIONS.png",
    TITLE=FIG_TITLE,
    CAPTION=PLOT_CAPTION_WRAPPED
)
```

```python
# Back-testing Early Warning System
BACKTEST_MONTHS = 12
BACKTEST_START_DATE = len(MONTHLY_EXPENSES_TIME_SERIES) - BACKTEST_MONTHS

MONTHLY_EXPENSES_TIME_SERIES_COPY = MONTHLY_EXPENSES_TIME_SERIES.copy()
```

```python
RESULTS = []
SARIMA_ORDER = (1,1,1)
SEASONAL_ORDER = (1,0,1,12)

for i in range(BACKTEST_START_DATE, len(MONTHLY_EXPENSES_TIME_SERIES_COPY)):
    TRAIN_DATA = MONTHLY_EXPENSES_TIME_SERIES_COPY.iloc[:i]

    MODEL = SARIMAX(
        TRAIN_DATA,
        order=SARIMA_ORDER,
        seasonal_order=SEASONAL_ORDER,
        enforce_stationarity=False,
        enforce_invertibility=False
    )

    FITTED_MODEL = MODEL.fit(
        disp=False
    )

    FORECAST = FITTED_MODEL.get_forecast(steps=1)

    MEAN = FORECAST.predicted_mean.iloc[0]
    CI = FORECAST.conf_int().iloc[0]

    LOWER_CI = CI[0]
    UPPER_CI = CI[1]

    ACTUAL_VALUE = MONTHLY_EXPENSES_TIME_SERIES_COPY.iloc[i]

    ANOMALY_CI = (ACTUAL_VALUE < LOWER_CI) or (ACTUAL_VALUE > UPPER_CI)

    RESULTS.append({
        "Date": MONTHLY_EXPENSES_TIME_SERIES_COPY.index[i],
        "Actual": ACTUAL_VALUE,
        "Forecast": MEAN,
        "Lower CI": LOWER_CI,
        "Upper CI": UPPER_CI,
        "Anomaly CI": ANOMALY_CI,
        "Error": ACTUAL_VALUE - MEAN
    })

BACKTEST_RESULTS = pd.DataFrame(RESULTS)
BACKTEST_RESULTS.set_index("Date", inplace=True)
BACKTEST_RESULTS
```

```python
# Define alert system based on SARIMA CI + historical patterns
HISTORICAL_MEAN = MONTHLY_EXPENSES_TIME_SERIES_COPY.mean()
HISTORICAL_STD = MONTHLY_EXPENSES_TIME_SERIES_COPY.std()

# Get average CI width from backtesting as basis for thresholds
AVG_CI_WIDTH = (BACKTEST_RESULTS["Upper CI"] - BACKTEST_RESULTS["Lower CI"]).mean()
AVG_FORECAST = BACKTEST_RESULTS["Forecast"].mean()

# Define four-tier alert thresholds
ALERT_SYSTEM = {
    "GREEN": {
        "Label": "Normal",
        "Condition": "Within 95% CI",
        "Upper Multiplier": 1.0,   # At CI boundary
        "Lower Multiplier": 1.0,
        "Color": "#4CAF50",
        "Action": "No action needed"
    },
    "YELLOW": {
        "Label": "Caution",
        "Condition": "1.0x - 1.25x beyond CI",
        "Upper Multiplier": 1.25,
        "Lower Multiplier": 1.25,
        "Color": "#FFC107",
        "Action": "Monitor closely, review spending categories"
    },
    "ORANGE": {
        "Label": "Warning",
        "Condition": "1.25x - 1.5x beyond CI",
        "Upper Multiplier": 1.50,
        "Lower Multiplier": 1.50,
        "Color": "#FF9800",
        "Action": "Investigate causes, reduce discretionary spending"
    },
    "RED": {
        "Label": "Critical",
        "Condition": "Beyond 1.5x CI",
        "Upper Multiplier": None,   # Anything beyond orange
        "Lower Multiplier": None,
        "Color": "#F44336",
        "Action": "URGENT - Immediate budget review required"
    }
}

def GET_ALERT_THRESHOLDS(FORECAST_MEAN, CI_LOWER, CI_UPPER):
    """Calculate all four alert tier boundaries for a given forecast."""
    CI_HALF_WIDTH_UPPER = CI_UPPER - FORECAST_MEAN
    CI_HALF_WIDTH_LOWER = FORECAST_MEAN - CI_LOWER

    return {
        # Upper thresholds
        "Upper Green": CI_UPPER,
        "Upper Yellow": CI_UPPER + (CI_HALF_WIDTH_UPPER * 0.25),
        "Upper Orange": CI_UPPER + (CI_HALF_WIDTH_UPPER * 0.50),
        # Lower thresholds
        "Lower Green": CI_LOWER,
        "Lower Yellow": CI_LOWER - (CI_HALF_WIDTH_LOWER * 0.25),
        "Lower Orange": CI_LOWER - (CI_HALF_WIDTH_LOWER * 0.50),
        # Reference
        "Forecast": FORECAST_MEAN,
        "CI Upper": CI_UPPER,
        "CI Lower": CI_LOWER
    }

def CLASSIFY_ALERT(actual, thresholds):
    """Classify a spending value into an alert tier."""
    if thresholds["Lower Green"] <= actual <= thresholds["Upper Green"]:
        return "GREEN"
    elif thresholds["Lower Yellow"] <= actual <= thresholds["Upper Yellow"]:
        return "YELLOW"
    elif thresholds["Lower Orange"] <= actual <= thresholds["Upper Orange"]:
        return "ORANGE"
    else:
        return "RED"

# Visualize the threshold system conceptually
# Use the average forecast as a reference point
REF_THRESHOLDS = GET_ALERT_THRESHOLDS(
    AVG_FORECAST,
    AVG_FORECAST - AVG_CI_WIDTH / 2,
    AVG_FORECAST + AVG_CI_WIDTH / 2
)

# Build threshold definition table
THRESHOLD_TABLE = pd.DataFrame({
    "Alert Level": ["🟢 GREEN", "🟡 YELLOW", "🟠 ORANGE", "🔴 RED"],
    "Label": ["Normal", "Caution", "Warning", "Critical"],
    "Condition": [
        "Within 95% CI",
        "1.0x - 1.25x beyond CI",
        "1.25x - 1.5x beyond CI",
        "Beyond 1.5x CI"
    ],
    "Upper Bound ($)": [
        f"${REF_THRESHOLDS["Upper Green"]:.2f}",
        f"${REF_THRESHOLDS["Upper Yellow"]:.2f}",
        f"${REF_THRESHOLDS["Upper Orange"]:.2f}",
        "Above orange"
    ],
    "Lower Bound ($)": [
        f"${REF_THRESHOLDS["Lower Green"]:.2f}",
        f"${REF_THRESHOLDS["Lower Yellow"]:.2f}",
        f"${REF_THRESHOLDS["Lower Orange"]:.2f}",
        "Below orange"
    ],
    "Recommended Action": [
        details["Action"] for details in ALERT_SYSTEM.values()
    ]
})

# Plot threshold definition as visual table
fig = go.Figure(data=[go.Table(
    header=dict(
        values=[f"<b>{col}</b>" for col in THRESHOLD_TABLE.columns],
        fill_color="#1e293b",
        align="left",
        font=dict(color="white", size=12),
        height=35
    ),
    cells=dict(
        values=[THRESHOLD_TABLE[col] for col in THRESHOLD_TABLE.columns],
        fill_color=[
            ["#1a3a2a", "#3a3010", "#3a2010", "#3a1010"],  # Alert Level
            ["#1a3a2a", "#3a3010", "#3a2010", "#3a1010"],  # Label
            "#334155",  # Condition
            "#334155",  # Upper
            "#334155",  # Lower
            "#334155",  # Action
        ],
        align="left",
        font=dict(
            color=[
                ["#4CAF50", "#FFC107", "#FF9800", "#F44336"],
                ["#4CAF50", "#FFC107", "#FF9800", "#F44336"],
                "white", "white", "white", "white"
            ],
            size=11
        ),
        height=35
    )
)])

fig.update_layout(
    title=dict(
        text="Early Warning System - Alert Threshold Definitions",
        x=0.5,
        xanchor="center",
        font=dict(size=18)
    ),
    height=450,
    template="plotly_dark"
)

fig.show()

FIG_TITLE = "Early Warning System - Alert Threshold Definitions"
PLOT_CAPTION = f"""
The early warning system defines four alert tiers based on SARIMA confidence interval boundaries and historical spending patterns. The GREEN zone represents normal spending within the 95% confidence interval, requiring no intervention. YELLOW caution alerts trigger when spending deviates 1.0-1.25x beyond CI bounds, signalling the need for closer monitoring. ORANGE warnings indicate moderate anomalies (1.25-1.5x beyond CI) requiring active investigation and corrective action. RED critical alerts fire for severe deviations beyond 1.5x CI boundaries, demanding immediate budget review. Reference thresholds are calculated from the average forecast of ${AVG_FORECAST:.2f} and average CI width of ${AVG_CI_WIDTH:.2f}, with dynamic thresholds recalculated monthly based on current SARIMA forecasts.
"""

PLOT_CAPTION_WRAPPED = textwrap.fill(PLOT_CAPTION, width=130)
print(PLOT_CAPTION_WRAPPED)
print("\n--------------------------------\n")

SAVE_PLOT_AND_REGISTER(
    fig,
    FILENAME="EARLY_WARNING_THRESHOLD_DEFINITION.png",
    TITLE=FIG_TITLE,
    CAPTION=PLOT_CAPTION_WRAPPED
)
```

```python
# Use backtest results for error analysis
ERROR_ANALYSIS = BACKTEST_RESULTS.copy()

# Calculate comprehensive error metrics
ERROR_METRICS = {
    # Bias (systematic over/under prediction)
    "MEAN_ERROR": ERROR_ANALYSIS["Error"].mean(),
    "MEDIAN_ERROR": ERROR_ANALYSIS["Error"].median(),
    
    # Variance (consistency)
    "STD_ERROR": ERROR_ANALYSIS["Error"].std(),
    "VARIANCE": ERROR_ANALYSIS["Error"].var(),
    
    # Accuracy
    "MAE": ERROR_ANALYSIS["Error"].abs().mean(),
    "RMSE": np.sqrt((ERROR_ANALYSIS["Error"]**2).mean()),
    "MAPE": (ERROR_ANALYSIS["Error"].abs() / ERROR_ANALYSIS["Actual"]).mean() * 100,
    
    # Distribution
    "SKEWNESS": ERROR_ANALYSIS["Error"].skew(),
    "KURTOSIS": ERROR_ANALYSIS["Error"].kurtosis(),
    
    # Extremes
    "MAX_OVERPREDICTION": ERROR_ANALYSIS["Error"].min(),  # Negative = underspent vs prediction
    "MAX_UNDERPREDICTION": ERROR_ANALYSIS["Error"].max()  # Positive = overspent vs prediction
}

ERROR_METRICS
```

```python
# Bias Classification
if abs(ERROR_METRICS["MEAN_ERROR"]) < ERROR_METRICS["STD_ERROR"] * 0.1:
    BIAS_ASSESSMENT = "UNBIASED - No systematic over/under prediction"
elif ERROR_METRICS["MEAN_ERROR"] > 0:
    BIAS_ASSESSMENT = f"POSITIVE BIAS - Tends to underpredict by ${ERROR_METRICS['MEAN_ERROR']:.2f}/month"
else:
    BIAS_ASSESSMENT = f"NEGATIVE BIAS - Tends to overpredict by ${abs(ERROR_METRICS['MEAN_ERROR']):.2f}/month"
```

```python
 # Coefficient of Variation
CV = ERROR_METRICS['STD_ERROR'] / ERROR_METRICS['MAE'] 
if CV < 0.5:
    CONSISTENCY = "HIGH - Errors are consistent in magnitude"
elif CV < 1.0:
    CONSISTENCY = "MODERATE - Some variability in error magnitude"
else:
    CONSISTENCY = "LOW - High variability in prediction accuracy"
```

```python
# Error Attribution & Failure Analysis
ERROR_ANALYSIS["Error Category"] = pd.cut(
    ERROR_ANALYSIS["Abs Error"],
    bins=[0, 50, 100, 200, float("inf")],
    labels=["Small (<$50)", "Medium ($50-$100)", "Large ($100-$200)", "Extreme (>$200)"]
)

# Count errors by category
ERROR_CATEGORY_COUNTS = ERROR_ANALYSIS["Error Category"].value_counts().sort_index()

# Identify top error contributors
TOP_ERRORS = ERROR_ANALYSIS.nlargest(5, "Abs Error")[
    ["Actual", "Forecast", "Error", "Abs Error", "Lower CI", "Upper CI"]
].copy()

TOP_ERRORS["Deviation PCT"] = (TOP_ERRORS["Error"] / TOP_ERRORS["Forecast"] * 100).round(1)

# Calculate error concentration
TOTAL_SQUARED_ERROR = (ERROR_ANALYSIS["Error"] ** 2).sum()
ERROR_ANALYSIS["Squared Error"] = ERROR_ANALYSIS["Error"] ** 2
ERROR_ANALYSIS["Contribution PCT"] = (
    ERROR_ANALYSIS["Squared Error"] / TOTAL_SQUARED_ERROR * 100
).round(2)

# Find months contributing most to total error
TOP_CONTRIBUTORS = ERROR_ANALYSIS.nlargest(5, "Contribution PCT")[
    ["Actual", "Forecast", "Error", "Abs Error", "Contribution PCT"]
]

# Pareto analysis: What % of months account for 80% of total squared error?
ERROR_SORTED = ERROR_ANALYSIS.sort_values("Squared Error", ascending=False).copy()
ERROR_SORTED["Cumulative PCT"] = ERROR_SORTED["Contribution PCT"].cumsum()
MONTHS_FOR_80PCT = (ERROR_SORTED["Cumulative PCT"] <= 80).sum()
PARETO_PCT = MONTHS_FOR_80PCT / len(ERROR_ANALYSIS) * 100

print("="*70)
print("ERROR ATTRIBUTION & FAILURE ANALYSIS")
print("="*70)

print(f"\n📊 ERROR MAGNITUDE DISTRIBUTION:")
for category, count in ERROR_CATEGORY_COUNTS.items():
    pct = count / len(ERROR_ANALYSIS) * 100
    print(f"  {category:20s}: {count:2d} months ({pct:5.1f}%)")

print(f"\n🔥 TOP 5 LARGEST ERRORS:")
print("-" * 70)
for idx, row in TOP_ERRORS.iterrows():
    print(f"  {idx.strftime("%B %Y")}:")
    print(f"    Actual: ${row["Actual"]:.2f} | Forecast: ${row["Forecast"]:.2f}")
    print(f"    Error: ${row["Error"]:+.2f} ({row["Deviation PCT"]:+.1f}%)")
    print(f"    CI: [${row["Lower CI"]:.2f}, ${row["Upper CI"]:.2f}]")
    print()

print(f"📈 ERROR CONCENTRATION (Pareto Analysis):")
print(f"  {MONTHS_FOR_80PCT} months ({PARETO_PCT:.1f}%) account for 80% of total squared error")
print(f"  This suggests errors are {"concentrated" if PARETO_PCT < 50 else "distributed"} across the back-testing period")

print(f"\n🎯 TOP 5 ERROR CONTRIBUTORS (by squared error):")
print("-" * 70)
for idx, row in TOP_CONTRIBUTORS.iterrows():
    print(f"  {idx.strftime("%B %Y")}: {row["Contribution PCT"]:.1f}% of total error")
    print(f"    Error: ${row["Error"]:+.2f}, Abs Error: ${row["Abs Error"]:.2f}")

print("="*70)
```

```python
# Analyze CI performance
ERROR_ANALYSIS['Within CI'] = (
    (ERROR_ANALYSIS['Actual'] >= ERROR_ANALYSIS['Lower CI']) &
    (ERROR_ANALYSIS['Actual'] <= ERROR_ANALYSIS['Upper CI'])
)

ERROR_ANALYSIS['CI Width'] = ERROR_ANALYSIS['Upper CI'] - ERROR_ANALYSIS['Lower CI']
ERROR_ANALYSIS['CI Width PCT'] = (
    ERROR_ANALYSIS['CI Width'] / ERROR_ANALYSIS['Forecast'] * 100
)

# Calculate coverage statistics
COVERAGE_RATE = ERROR_ANALYSIS['Within CI'].mean() * 100
EXPECTED_COVERAGE = 95  # 95% CI

# Violations breakdown
UPPER_VIOLATIONS = (ERROR_ANALYSIS['Actual'] > ERROR_ANALYSIS['Upper CI']).sum()
LOWER_VIOLATIONS = (ERROR_ANALYSIS['Actual'] < ERROR_ANALYSIS['Lower CI']).sum()
TOTAL_VIOLATIONS = UPPER_VIOLATIONS + LOWER_VIOLATIONS

# CI width statistics
AVG_CI_WIDTH = ERROR_ANALYSIS['CI Width'].mean()
MIN_CI_WIDTH = ERROR_ANALYSIS['CI Width'].min()
MAX_CI_WIDTH = ERROR_ANALYSIS['CI Width'].max()

# Find periods with narrowest and widest CIs
NARROWEST_CI_MONTH = ERROR_ANALYSIS.loc[ERROR_ANALYSIS['CI Width'].idxmin()]
WIDEST_CI_MONTH = ERROR_ANALYSIS.loc[ERROR_ANALYSIS['CI Width'].idxmax()]

# Analyze relationship between CI width and errors
ERROR_ANALYSIS['CI Violation'] = ~ERROR_ANALYSIS['Within CI']

print("="*70)
print("CONFIDENCE INTERVAL COVERAGE ANALYSIS")
print("="*70)

print(f"\n📊 COVERAGE STATISTICS:")
print(f"  Expected Coverage (95% CI):     {EXPECTED_COVERAGE:.1f}%")
print(f"  Actual Coverage:                {COVERAGE_RATE:.1f}%")
print(f"  Coverage {'meets' if COVERAGE_RATE >= 90 else 'below'} expectations")

print(f"\n⚠️  VIOLATIONS:")
print(f"  Total months outside CI:        {TOTAL_VIOLATIONS} ({TOTAL_VIOLATIONS/len(ERROR_ANALYSIS)*100:.1f}%)")
print(f"  Upper bound violations:         {UPPER_VIOLATIONS} (overspending)")
print(f"  Lower bound violations:         {LOWER_VIOLATIONS} (underspending)")

print(f"\n📏 CONFIDENCE INTERVAL WIDTH:")
print(f"  Average CI Width:               ${AVG_CI_WIDTH:.2f}")
print(f"  As % of forecast:               {ERROR_ANALYSIS['CI Width PCT'].mean():.1f}%")
print(f"  Minimum Width:                  ${MIN_CI_WIDTH:.2f} ({NARROWEST_CI_MONTH.name.strftime('%B %Y')})")
print(f"  Maximum Width:                  ${MAX_CI_WIDTH:.2f} ({WIDEST_CI_MONTH.name.strftime('%B %Y')})")

# Calculate if violations occur when CI is narrow
VIOLATIONS_DF = ERROR_ANALYSIS[ERROR_ANALYSIS['CI Violation']]
if len(VIOLATIONS_DF) > 0:
    AVG_CI_VIOLATIONS = VIOLATIONS_DF['CI Width'].mean()
    AVG_CI_NO_VIOLATIONS = ERROR_ANALYSIS[~ERROR_ANALYSIS['CI Violation']]['CI Width'].mean()
    
    print(f"\n🔍 CI WIDTH vs VIOLATIONS:")
    print(f"  Avg CI width (violations):      ${AVG_CI_VIOLATIONS:.2f}")
    print(f"  Avg CI width (no violations):   ${AVG_CI_NO_VIOLATIONS:.2f}")
    if AVG_CI_VIOLATIONS < AVG_CI_NO_VIOLATIONS:
        print(f"  → Violations tend to occur when CI is narrower (overconfident predictions)")
    else:
        print(f"  → CI width does not predict violations well")

print("="*70)
```

```python
# Generate scenario forecasts
SCENARIO_FORECASTS = {}

for key, scenario in SCENARIOS.items():
    SCENARIO_DATA = BASELINE_DF.copy()
    
    # Apply adjustment
    if scenario["ADJUSTMENT"] != 0:
        SCENARIO_DATA["Forecast"] = SCENARIO_DATA["Forecast"] * (1 + scenario["ADJUSTMENT"])
        SCENARIO_DATA["Lower CI"] = SCENARIO_DATA["Lower CI"] * (1 + scenario["ADJUSTMENT"])
        SCENARIO_DATA["Upper CI"] = SCENARIO_DATA["Upper CI"] * (1 + scenario["ADJUSTMENT"])
    
    # Apply spike if exists
    if "SPIKE_MONTH" in scenario:
        spike_idx = scenario["SPIKE_MONTH"] - 1
        SCENARIO_DATA.iloc[spike_idx, SCENARIO_DATA.columns.get_loc("Forecast")] += scenario["SPIKE_AMOUNT"]
    
    # Calculate totals
    SCENARIO_DATA["Cumulative"] = SCENARIO_DATA["Forecast"].cumsum()
    
    SCENARIO_FORECASTS[key] = {
        "DATA": SCENARIO_DATA,
        "TOTAL": SCENARIO_DATA["Forecast"].sum(),
        "MONTHLY_AVG": SCENARIO_DATA["Forecast"].mean(),
        **scenario
    }

# Comparison summary
print("\n" + "="*70)
print("SCENARIO COMPARISON SUMMARY")
print("="*70)
print(f"{"Scenario":<35s} {"Annual Total":>15s} {"vs Baseline":>12s} {"Monthly Avg":>12s}")
print("-" * 70)

BASELINE_TOTAL = SCENARIO_FORECASTS["BASELINE"]["TOTAL"]

for key in ["BASELINE", "OPTIMIZATION", "INCOME_LOSS", "AUSTERITY", "LIFESTYLE_INFLATION", "MAJOR_EXPENSE"]:
    SCENARIO = SCENARIO_FORECASTS[key]
    DIFF = SCENARIO["TOTAL"] - BASELINE_TOTAL
    DIFF_PCT = (DIFF / BASELINE_TOTAL * 100) if BASELINE_TOTAL != 0 else 0
    
    print(f"{SCENARIO["NAME"]:<35s} ${SCENARIO["TOTAL"]:>13,.2f} {DIFF:+>11,.2f} ({DIFF_PCT:+.1f}%)  ${SCENARIO["MONTHLY_AVG"]:>10,.2f}")

print("="*70)
```
